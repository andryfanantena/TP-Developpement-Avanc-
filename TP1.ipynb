```python
#1. moyenne, max, 
import numpy as np
note = [12, 8, 15, 17, 9, 11, 14, 6, 13, 16]
print (len(note),"notes")
print ("moyenne :",np.mean(note))
print ("max : ", max(note))
print ("min : ", min(note))
```

```python
#2. liste 
admis = [n for n in note if n >= 10]
print (admis)
```

```python
#3. ordre croissant
print("croissant : ", sorted(note))
```

```python
#3.3 meilleurs notes
notes=(sorted(note, reverse=True))
print ("3 Meilleures notes: ",notes[:3])
ajournes = len([n for n in notes if n < 10])
print("ajournés :", ajournes)
```

```python
#4. nouvelle liste
nouvelle_notes = [min(10, n + 2) if n < 10 else n for n in note]
print("nouvelle notes :", nouvelle_notes)
```

```python
#5. fonction histogramme
def histogramme(notes):
    colonne = [(0, 4), (5, 9), (10, 14), (15, 20)]
    for min, max in colonne:
        x = sum(1 for n in notes if min <= n <= max)
        print(f"{min:>2}-{max:<2} | {'*' * x}")
print("----------Histogramme----------")
histogramme(notes)
```

```python
#6. focntion rang
def rang(notes, note):
    classement = sorted(notes, reverse=True)
    try:
        return classement.index(note) + 1
    except ValueError:
        return None
print(f"Rang de 14 : {rang(notes, 14)}")
```
