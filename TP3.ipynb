```python
# 1. Affichage des produits, prix et quantités
import json
from collections import Counter
stock = {
    "riz":   {"prix": 3200, "quantite": 50},
    "huile": {"prix": 9500, "quantite": 12},
    "sucre": {"prix": 4000, "quantite": 30},
    "savon": {"prix": 1500, "quantite": 0},
}
for produit, informations in stock.items():
    print(f"{produit} : {informations['prix']} Ar, "
          f"quantité = {informations['quantite']}")
```

```python
# 2. Valeur totale du stock
valeur_totale = sum(
    infos["prix"] * infos["quantite"]
    for infos in stock.values()
)
print("Valeur totale :", f"{valeur_totale:,}".replace(",", " "), "Ar")
```

```python
# 3. Produits en rupture
rupture = [nom for nom, infos in stock.items() if infos["quantite"] == 0]
print(rupture)
```

```python
# 4. Vendre un produit
def vendre(stock, produit, quantite):
    if quantite <= 0:
        return False, "La quantité à vendre doit être positive."
    infos = stock.get(produit)
    if infos is None:
        return False, f"Produit inconnu : {produit}."
    if infos["quantite"] < quantite:
        return False, f"Stock insuffisant pour {produit}."
    infos["quantite"] -= quantite
    return True, f"Vente effectuée : {quantite} unité(s) de {produit}."
print(vendre(stock, "riz", 5))
print(vendre(stock, "produit_inconnu", 2))
print(vendre(stock, "savon", 1))
```

```python
# 5. Ajouter un produit en refusant les doublons
def ajouter_produit(stock, nom, prix, quantite):
    if nom in stock:
        return False, f"Le produit {nom} existe déjà."
    if prix < 0 or quantite < 0:
        return False, "Le prix et la quantité doivent être positifs ou nuls."
    stock[nom] = {"prix": prix, "quantite": quantite}
    return True, f"Produit {nom} ajouté."
print(ajouter_produit(stock, "lait", 4500, 8))
print(ajouter_produit(stock, "riz", 3200, 10))
```

```python
# 6. HAusse de 10%
prix_hausse = {nom: round(infos["prix"] * 1.10 )for nom, infos in stock.items()}
print(f"Hausse de 10 % : {prix_hausse}")
```

```python
# 7. Compter les ventes avec get, puis Counter
ventes = ["riz", "sucre", "riz", "huile", "riz"]
compteur = {}
for produit in ventes:
    compteur[produit] = compteur.get(produit, 0) + 1
print("Get :", compteur)
print("Collection Counter :", dict(Counter(ventes)))
```

```python
# 8. Sauvegarder et recharger le stock en JSON
nom_fichier = "stock.json"
with open(nom_fichier, "w", encoding="utf-8") as fichier:
    json.dump(stock, fichier, ensure_ascii=False, indent=2)
with open(nom_fichier, "r", encoding="utf-8") as fichier:
    stock_recharge = json.load(fichier)
print("Sauvegarde :", nom_fichier)
print("Rechargement :", stock == stock_recharge)
```
