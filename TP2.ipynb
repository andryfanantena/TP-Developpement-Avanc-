```python
# 1. les tuples
from math import dist, radians, sin, cos, asin, sqrt
from itertools import combinations
from collections import namedtuple
villes = [
    ("Antananarivo", -18.88, 47.51),
    ("Toamasina", -18.15, 49.40),
    ("Mahajanga", -15.72, 46.32),
    ("Fianarantsoa", -21.45, 47.09),
]
for nom, latitude, longitude in villes:
    print(f"{nom} : {latitude}, {longitude}")
```

```python
# 2. Ville la plus au nord
Nord = max(villes, key=lambda ville: ville[1])
print(Nord[0])
```

```python
# 3. Fonction bornes : lat_min, lat_max, lon_min, lon_max
def bornes(villes):
    lat = [ville[1] for ville in villes]
    long = [ville[2] for ville in villes]
    return min(lat), max(lat), min(long), max(long)
lat_min, lat_max, lon_min, lon_max = bornes(villes)
print("Latitude max :", lat_max)
print("Latitude min :", lat_min)
print("Longitude min :", lon_max)
print("Longitude max :", lon_max)
```

```python
# 4. Distance euclidienne en degrés
def distance(p1, p2):
    return dist(p1, p2)
point_tana = (villes[0][1], villes[0][2])
point_toamasina = (villes[1][1], villes[1][2])
print(round(distance(point_tana, point_toamasina), 4))
```

```python
# 5. Dictionnaire des distances pour toutes les paires de villes
distances = {}
for ville1, ville2 in combinations(villes, 2):
    cle = (ville1[0], ville2[0])
    coord1 = (ville1[1], ville1[2])
    coord2 = (ville2[1], ville2[2])
    distances[cle] = distance(coord1, coord2)
for paire, valeur in distances.items():
    print(f"{paire} : {valeur:.4f} degré(s)")
print("Nombre de paires :", len(distances))
```

```python
# 6. Un tuple ne peut pas être modifié : créer un nouveau tuple
try:
    villes[0][1] = -18.90
except TypeError:
    print("Erreur !!!")
    print("Un tuple est non modifiable. Il faut créer un nouveau tuple.")
ancienne_ville = villes[0]
villes[0] = (ancienne_ville[0], -18.90, ancienne_ville[2])
print("\nVille corrigée :", villes[0])
```

```python
# 7. Bonus : namedtuple et distance réelle avec la formule de Haversine
Ville = namedtuple("Ville", ["nom", "latitude", "longitude"])
villes_nommees = [Ville(*ville) for ville in villes]

def distance_km(ville1, ville2):
    rayon_terre = 6371.0
    lat1, lon1 = radians(ville1.latitude), radians(ville1.longitude)
    lat2, lon2 = radians(ville2.latitude), radians(ville2.longitude)
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat / 2) ** 2 + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    return 2 * rayon_terre * asin(sqrt(a))

print("distance approximative")
print(f"{villes_nommees[0].nom} - {villes_nommees[1].nom} : "
      f"{distance_km(villes_nommees[0], villes_nommees[1]):.2f} km")
```
